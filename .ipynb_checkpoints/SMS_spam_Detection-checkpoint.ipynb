## 3.Text preprocessing
* lower case
* Tokenization
* Removing special characters
* Removing stopwords and punctuations
* Stemming

```python
# creating a function that do Text preprocessing complete together
from nltk.corpus import stopwords
import string
from nltk.stem.porter import PorterStemmer
PS = PorterStemmer()
stopwords = set(stopwords.words('english'))
punctuation = set(string.punctuation)

def text_transform(text):
    text = text.lower() #lower case
    text = nltk.word_tokenize(text) #tokenize

    y = []
    # alphanumeric or remove special chars
    for i in text:
        if i.isalnum():
            y.append(i)
    text = y.copy()
    y.clear()

    #remove stopwords and punctuation
    for i in text:
        if i not in stopwords and i not in punctuation:
            y.append(i)
    text = y.copy()
    y.clear()

    #stemming
    for i in text:
        y.append(PS.stem(i))
    text = y.copy()
    y.clear()
            
    return " ".join(text)
```

```python
df['transformed_text'] = df['text'].apply(text_transform)
```

## 4.Building Model

* <h1 style="color:#0077b6;">Bag of word using CountVectorizer</h1>

```python


from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
cv = CountVectorizer()
```

```python
X = cv.fit_transform(df['transformed_text']).toarray()
X.shape
```

```python
y = df['target'].values
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2, random_state = 2)
```

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score
```

```python
gnb = GaussianNB()
mnb = MultinomialNB()
bnb = BernoulliNB()
```

```python
gnb_model = gnb.fit(X_train,y_train)
y_pred_gnb = gnb_model.predict(X_test)
print("---------------------------- For Gaussian NB Model ----------------------------")
print(f"Test Accuracy of gnb model : {accuracy_score(y_test, y_pred_gnb)}")
print(f"Test Precision of gnb model : {precision_score(y_test, y_pred_gnb)}")
print(f"Test Confusion Matrix of gnb model : {confusion_matrix(y_test, y_pred_gnb)}")
print('\n')

print("---------------------------- For Multinomial NB Model ----------------------------")
mnb_model = mnb.fit(X_train,y_train)
y_pred_mnb = mnb_model.predict(X_test)
print(f"Test accuracy of mnb model : {accuracy_score(y_test, y_pred_mnb)}")
print(f"Test Precision of mnb model : {precision_score(y_test, y_pred_mnb)}")
print(f"Test Confusion Matrix of mnb model : {confusion_matrix(y_test, y_pred_mnb)}")
print('\n')


print("---------------------------- For Bernaulli NB Model ----------------------------")
bnb_model = bnb.fit(X_train,y_train)
y_pred_bnb = bnb_model.predict(X_test)
print(f"Test accuracy of bnb model : {accuracy_score(y_test, y_pred_bnb)}")
print(f"Test Precision of bnb model : {precision_score(y_test, y_pred_bnb)}")
print(f"Test Confusion Matrix of bnb model : {confusion_matrix(y_test, y_pred_bnb)}")
```

## Now use other models incluse Ensamble learning

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    BaggingClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)
from xgboost import XGBClassifier

```

```python
svc = SVC(kernel='sigmoid', gamma=1.0)
knc = KNeighborsClassifier()
mnb = MultinomialNB()
dtc = DecisionTreeClassifier(max_depth=5)
lrc = LogisticRegression(solver='liblinear', penalty='l1')
rfc = RandomForestClassifier(n_estimators=50, random_state=2)
abc = AdaBoostClassifier(n_estimators=50, random_state=2)
bc = BaggingClassifier(n_estimators=50, random_state=2)
etc = ExtraTreesClassifier(n_estimators=50, random_state=2)
gbdt = GradientBoostingClassifier(n_estimators=50, random_state=2)
xgb = XGBClassifier(n_estimators=50, random_state=2, eval_metric='logloss')
```

```python
clfs = {
    'SVC' : svc,
    'KN' : knc,
    'NB' :mnb,
    'DT' : dtc,
    'LR' : lrc,
    'RF' : rfc,
    'AdaBoost' : abc,
    'ETC' : etc,
    'GBDT' : gbdt,
    'xgb' : xgb,
    'BgC' : bc
}
```

```python
def train_classifier(clf, X_train,y_train, X_test, y_test):
    clf.fit(X_train,y_train)
    y_pred = clf.predict(X_test)
    precision = precision_score(y_test, y_pred)
    accuracy = accuracy_score(y_test, y_pred)

    return accuracy, precision
    
```

```python
# train_classifier(bc, X_train, y_train, X_test, y_test)
```

```python
acc_score = []
prec_score = []
results = []

def ALL_Model(clfs, X_train, y_train, X_test, y_test):
    for name,clf in clfs.items():
        current_acc_score, current_prec_score = train_classifier(clf,X_train, y_train, X_test, y_test)  #FUNCTION CALL
        print(f"For {name}")
        print(f"Accuracy - {current_acc_score}")
        print(f"Pecision - {current_prec_score}\n")
        
        results.append({
            'Algorithm': name,
            'Accuracy': current_acc_score,
            'Precision': current_prec_score
        })

    compare_performance = pd.DataFrame(results)
    
    return compare_performance.sort_values(
        'Precision', ascending=False
    ).reset_index(drop=True)
 
    
```

```python
compare_performance = ALL_Model(clfs,X_train, y_train, X_test, y_test)
```

```python
plt.figure(figsize=(12, 6))

compare_performance = compare_performance.melt(
    id_vars='Algorithm',
    value_vars=['Accuracy', 'Precision'],
    var_name='Metric',
    value_name='Score'
)

sns.barplot(
    data=compare_performance,
    x='Algorithm',
    y='Score',
    hue='Metric'
)
```
